*validate_schema(df, expected_columns)* - sprawdza, czy DataFrame zawiera wszystkie wymagane kolumny, zwraca błąd z informacją, które brakuje  

*safe_read_json(path)* - ładuje plik JSON i przechwytuje wyjątki  

*log_pipeline_step(step_name)* każdy etap pipeline loguje swój status, pomaga w debugowaniu i analizie błędów  

*enforce_value_ranges(df, column, min_val, max_val)* - automatyczna walidacja wartości  

*auto_backup_output(path, df)* - zanim pipeline nadpisze plik wynikowy, robi jego kopię z timestampem
